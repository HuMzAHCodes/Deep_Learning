```python
def build_model(hp):
  model = Sequential()

  # hp.Int needs a name (for tracking), then min_value, max_value, step
  # this tells the tuner to try unit counts from 8 to 128, in steps of 8
  # (i.e. 8, 16, 24, 32, ..., 128) and search for the best one
  units = hp.Int('units', min_value=8, max_value=128, step=8)

  model.add(Dense(units=units, activation="relu", input_dim=8))
  model.add(Dense(1, activation="sigmoid"))

  model.compile(optimizer="rmsprop", loss="binary_crossentropy", metrics=["accuracy"])

  return model
```

```python
tuner = kt.RandomSearch(build_model, objective="val_accuracy", max_trials=5,directory="mydir",project_name="hamza")
```

```python
tuner.search(X_train, Y_train, epochs=5, validation_data=(X_test, Y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
def build_model(hp):
  model = Sequential()

  # first (input) layer — fixed size, always present
  model.add(Dense(80, activation="relu", input_dim=8))

  # hp.Int lets the tuner search how many hidden layers to add,
  # trying anywhere from 1 to 10 layers
  for i in range(hp.Int("num_layers", min_value=1, max_value=10)):
    # add one hidden layer per loop iteration
    model.add(Dense(80, activation="relu"))

  # output layer — single neuron for binary classification
  model.add(Dense(1, activation="sigmoid"))

  model.compile(optimizer="rmsprop", loss="binary_crossentropy", metrics=["accuracy"])

  return model
```

```python
tuner = kt.RandomSearch(
    build_model,                  # the model-building function to search over
    objective="val_accuracy",     # metric the tuner tries to optimize
    max_trials=4,                 # number of different hyperparameter combos to try
    directory="mydir",            # folder to store search results/logs
    project_name="num_layers"     # subfolder name for this specific search
)
```

```python
tuner.search(
    X_train, Y_train,             # training data and labels
    epochs=5,                     # number of epochs to train each trial for
    validation_data=(X_test, Y_test)  # data used to evaluate each trial's performance
)
```

```python
tuner.get_best_hyperparameters()[0].values
# this is for the recent task ( number of layers)
```

```python
model=tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(X_train, Y_train, epochs=100, initial_epoch=6, validation_data=(X_test, Y_test))
```

```python
def build_model(hp):
  model = Sequential()

  counter = 0

  # hp.Int('num_layers', min_value=1, max_value=10) searches how many
  # hidden layers to add (must be INSIDE hp.Int, not passed to range())
  for i in range(hp.Int('num_layers', min_value=1, max_value=10)):

    if counter == 0:
      # first hidden layer — needs input_dim since it's right after the input
      model.add(
          Dense(
              hp.Int('units' + str(i), min_value=8, max_value=128, step=8),
              activation=hp.Choice('activation' + str(i), values=['relu', 'tanh', 'sigmoid']),
              input_dim=8
          )
      )

      model.add(Dropout(hp.Choice('dropout' + str(i), values=[.1, .2, .3, .4, .5, .6, .7])))
    else:
      # subsequent hidden layers — no input_dim needed, Keras infers shape
      model.add(
          Dense(
              hp.Int('units' + str(i), min_value=8, max_value=128, step=8),
              activation=hp.Choice('activation' + str(i), values=['relu', 'tanh', 'sigmoid'])
          )
      )
      model.add(Dropout(hp.Choice('dropout' + str(i), values=[.1, .2, .3, .4, .5, .6, .7])))
    counter += 1

  # output layer — single neuron for binary classification
  model.add(Dense(1, activation='sigmoid'))

  model.compile(
      optimizer=hp.Choice('optimizer', values=['rmsprop', 'adam', 'sgd', 'nadam', 'adadelta']),
      loss='binary_crossentropy',
      metrics=['accuracy']
  )

  return model
```

```python
tuner = kt.RandomSearch(
    build_model,                  # function that builds and compiles the model
    objective="val_accuracy",     # metric the tuner tries to optimize
    max_trials=3,                 # number of different hyperparameter combos to try
    directory="mydir",            # folder to store search results/logs
    project_name="final1"          # subfolder name for this specific search
)
```

```python
tuner.search(X_train,Y_train,epochs=5,validation_data=(X_test,Y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model=tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(X_train, Y_train, epochs=200, initial_epoch=6, validation_data=(X_test, Y_test))
```
