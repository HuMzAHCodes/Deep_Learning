```python
"""
OVERFITTING — cause, effect, and the full toolbox of fixes

Cause: a model with too much capacity relative to the amount/variety
of training data starts memorizing individual examples (including
their noise) instead of learning the general pattern. Small datasets
make this worse since there's less signal to average out over.

Effect: training loss/accuracy keeps improving while validation
loss/accuracy stalls or gets worse — a growing gap between train and
val performance is the signature symptom.

Ways to deal with it fall into two families:

1) INCREASE THE DATA — give the model more to generalize from, so
   memorization becomes harder relative to genuine pattern-learning.
     - Collect/add more rows of real data.
     - Generate more data (data augmentation / synthetic samples) when
       collecting new real data isn't feasible.

2) DECREASE THE MODEL'S EFFECTIVE COMPLEXITY — limit how much the
   model can memorize, forcing it to rely on more general patterns.
     - Dropout: randomly deactivate neurons during training so no
       single neuron/path can be over-relied on.
     - Regularization (L1/L2): penalize large weights in the loss
       function, discouraging the model from fitting noise.
     - Early stopping: halt training once validation performance
       stops improving, before the model starts memorizing.

This notebook focuses on Dropout specifically — the other techniques
are covered in the companion regularization notebook.
"""
```

```python
# Visualize the raw data — confirms the two classes are entangled,
# meaning a model needs some nonlinearity to separate them
plt.scatter(X[:,0], X[:,1], c=y)
plt.show()
```

```python
# Keras building blocks, including Dropout which we'll use later
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.optimizers import Adam
```

```python
"""
BASELINE MODEL — deliberately overcapacity (128+128 neurons for only
~200 2D points) and trained for 500 epochs with no regularization.
This is expected to overfit — used as the "before" comparison.
"""
model = Sequential()

model.add(Dense(128, input_dim=2, activation="relu"))
model.add(Dense(128, activation="relu"))
model.add(Dense(1, activation="sigmoid"))

adam = Adam(learning_rate=0.01)
model.compile(loss='binary_crossentropy', optimizer=adam, metrics=['accuracy'])

history = model.fit(X, y, epochs=500, validation_split=0.2, verbose=1)
```

```python
# Decision-region plotting utility to visually inspect the boundary
from mlxtend.plotting import plot_decision_regions
```

```python
# Plot the decision boundary — an overfit model typically shows a
# jagged, overly complex boundary that hugs individual points
plot_decision_regions(X, y.astype('int'), clf=model, legend=2)
plt.xlim(-0.7, 0.5)
plt.ylim(-0.8, 0.8)
plt.show()
```

```python
# Loss curves — watch for train loss dropping while val loss rises:
# the classic overfitting signature
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
# Accuracy curves — same check: train accuracy near-perfect,
# val accuracy plateaued/worse confirms overfitting
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```

```python
"""
FIX — same architecture, but with Dropout(0.5) after each Dense layer.
Randomly zeroing 50% of activations each step prevents co-adaptation
of neurons and should close the train/val gap seen above.
"""
model = Sequential()

model.add(Dense(128, input_dim=2, activation="relu"))
model.add(Dropout(0.5))
model.add(Dense(128, activation="relu"))
model.add(Dropout(0.5))
model.add(Dense(1, activation="sigmoid"))

adam = Adam(learning_rate=0.01)
model.compile(loss='binary_crossentropy', optimizer=adam, metrics=['accuracy'])

history = model.fit(X, y, epochs=500, validation_split=0.2, verbose=1)
```

```python
# Decision boundary with dropout — should look smoother/simpler
# than the overfit version, since the model can no longer rely on
# any single neuron to memorize a point
plot_decision_regions(X, y.astype('int'), clf=model, legend=2)
plt.xlim(-0.7, 0.5)
plt.ylim(-0.8, 0.8)
plt.title('p = 0.5')
plt.show()
```

```python
# Loss curves post-dropout — train and val loss should track each
# other much more closely than the baseline
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
# Accuracy curves post-dropout — smaller train/val gap = less overfitting
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```
